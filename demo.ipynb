# AI-Recruiter — демо

Пошаговый прогон: генерация демо-данных → индексация → ранжирование кандидатов по вакансии.

```python
from ai_recruiter.data import generate, generate_vacancies
from ai_recruiter.embeddings import build_index, get_store, get_embedder
from ai_recruiter.pipeline import run_pipeline
from ai_recruiter.schema import Vacancy
```

## 1. Демо-данные

Воспроизводимая генерация резюме (фиксированный seed) и тестовых вакансий.

```python
manifest = generate(n=100, seed=42)
vacancy_paths = generate_vacancies()
print(f"Резюме: {len(manifest)}, вакансий: {len(vacancy_paths)}")
```

## 2. Индексация

Парсинг → чанкинг → эмбеддинг → запись в векторное хранилище.

```python
store = build_index(manifest=manifest)
print(f"Чанков в индексе: {store.count()}")
```

## 3. Ранжирование по вакансии

Векторный поиск отбирает топ-кандидатов, затем агенты (Технический Скринер + HR-Аналитик) строят карточки.

```python
vacancy = Vacancy.load(vacancy_paths[0])
reports = run_pipeline(vacancy, top_k=5, store=store)
for r in reports:
    print(f"{r.full_name:40s} {r.score:5.1f}%  {r.verdict}")
```
